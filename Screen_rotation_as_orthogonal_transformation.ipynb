```python
# ============================================================
# B.Tech Mathematics Mini Project
# Project: Screen Rotation as Orthogonal Transformation (F6)
# ============================================================

print("=" * 65)
print("       SCREEN ROTATION AS ORTHOGONAL TRANSFORMATION")
print("=" * 65)

print("""
Objective:
1. Construct a 90° rotation matrix.
2. Verify that the matrix is orthogonal: Q^T Q = I.
3. Apply the matrix to screen coordinates.
4. Verify preservation of distances.
5. Demonstrate quadratic-form invariance.
6. Show that rotating twice gives a 180° rotation.
7. Visualize portrait and landscape screen coordinates.
""")
```

```python
# ============================================================
# IMPORTS
# ============================================================

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Make NumPy output easier to read
np.set_printoptions(precision=4, suppress=True)

print("Libraries imported successfully.")
```

```python
# ============================================================
# MATHEMATICAL MODEL
# ============================================================

theta = np.radians(ANGLE_DEGREES)

Q = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])

print("Rotation matrix Q:")
print(Q)

print("\nMathematical formula:")
print("""
        [ cos(theta)  -sin(theta) ]
Q =     [                         ]
        [ sin(theta)   cos(theta) ]
""")
```

```python
# ============================================================
# ROTATION MATRIX
# ============================================================

def rotation_matrix(angle_degrees, direction="CCW"):

    angle = np.radians(angle_degrees)

    if direction.upper() == "CW":
        angle = -angle

    Q = np.array([
        [np.cos(angle), -np.sin(angle)],
        [np.sin(angle),  np.cos(angle)]
    ])

    return Q


Q = rotation_matrix(
    ANGLE_DEGREES,
    ROTATION_DIRECTION
)

print("Rotation Matrix Q:")
print(Q)
```

```python
# ============================================================
# USER CONFIGURATION
# ============================================================

PORTRAIT_WIDTH = 1080
PORTRAIT_HEIGHT = 1920

# Change this value to test any angle
ANGLE_DEGREES = 90

# "CCW" = anticlockwise
# "CW"  = clockwise
ROTATION_DIRECTION = "CCW"

# Point for demonstration
DEMO_POINT = np.array([200.0, 300.0])

# Benchmark points
TEST_POINTS = np.array([
    [0, 0],
    [200, 300],
    [-300, 400],
    [400, -250],
    [-200, -350]
], dtype=float)

print("Configuration")
print("-" * 45)
print(f"Portrait size      : {PORTRAIT_WIDTH} × {PORTRAIT_HEIGHT}")
print(f"Rotation angle     : {ANGLE_DEGREES}°")
print(f"Direction          : {ROTATION_DIRECTION}")
print(f"Demo point         : {DEMO_POINT}")
```

```python
# ============================================================
# SYMBOLIC MATHEMATICAL MODEL
# ============================================================

theta_sym = sp.pi / 2

Q_sym = sp.Matrix([
    [sp.cos(theta_sym), -sp.sin(theta_sym)],
    [sp.sin(theta_sym),  sp.cos(theta_sym)]
])

print("Exact symbolic rotation matrix:")
sp.pprint(Q_sym)
```

```python
# ============================================================
# ORTHOGONALITY TEST
# ============================================================

QTQ = Q.T @ Q
identity = np.eye(2)

print("Q^T Q =")
print(QTQ)

print("\nIdentity matrix I =")
print(identity)

print("\nIs Q^T Q = I?")
print(np.allclose(QTQ, identity))

print("\nTherefore, Q is an orthogonal matrix.")
```

```python
# ============================================================
# DETERMINANT
# ============================================================

det_Q = np.linalg.det(Q)

print("det(Q) =", det_Q)

if np.isclose(det_Q, 1):
    print("The determinant is +1, so this is a proper rotation.")
else:
    print("The determinant is not +1.")
```

```python
# ============================================================
# APPLY ROTATION TO TEST POINTS
# ============================================================

# Each row is a point, so use Q.T when multiplying the matrix.
ROTATED_POINTS = TEST_POINTS @ Q.T

print("Original and rotated coordinates")
print("-" * 60)

for original, rotated in zip(TEST_POINTS, ROTATED_POINTS):
    print(
        f"Original: ({original[0]:8.2f}, {original[1]:8.2f})"
        f"  -->  "
        f"Rotated: ({rotated[0]:8.2f}, {rotated[1]:8.2f})"
    )
```

```python

# ============================================================
# DISTANCE / LENGTH PRESERVATION
# ============================================================

original_lengths = np.linalg.norm(TEST_POINTS, axis=1)
rotated_lengths = np.linalg.norm(ROTATED_POINTS, axis=1)

print("Distance preservation test")
print("-" * 65)
print(f"{'Point':<10}{'Original':>15}{'Rotated':>15}{'Difference':>15}")

for i in range(len(TEST_POINTS)):
    difference = abs(original_lengths[i] - rotated_lengths[i])

    print(
        f"{i:<10}"
        f"{original_lengths[i]:>15.4f}"
        f"{rotated_lengths[i]:>15.4f}"
        f"{difference:>15.8f}"
    )

print("\nMaximum error:",
      np.max(np.abs(original_lengths - rotated_lengths)))

print("\nDistance preservation verified:",
      np.allclose(original_lengths, rotated_lengths))
```

```python
# ============================================================
# QUADRATIC FORM INVARIANCE
# ============================================================

I = np.eye(2)

def quadratic_form(x, A=I):
    """
    Calculate q(x) = x^T A x
    """
    return x.T @ A @ x

print("Quadratic form: q(x) = x^T I x = x1^2 + x2^2")
print("-" * 65)
print(f"{'Point':<10}{'q(original)':>18}{'q(rotated)':>18}")

for i in range(len(TEST_POINTS)):
    q_original = quadratic_form(TEST_POINTS[i])
    q_rotated = quadratic_form(ROTATED_POINTS[i])

    print(
        f"{i:<10}"
        f"{q_original:>18.4f}"
        f"{q_rotated:>18.4f}"
    )

print("\nQuadratic form is invariant:",
      all(
          np.isclose(
              quadratic_form(TEST_POINTS[i]),
              quadratic_form(ROTATED_POINTS[i])
          )
          for i in range(len(TEST_POINTS))
      ))
```

```python
# ============================================================
# ROTATING TWICE
# ============================================================

Q_twice = Q @ Q

Q_180 = np.array([
    [-1, 0],
    [ 0, -1]
], dtype=float)

print("Q × Q =")
print(Q_twice)

print("\nExpected 180° rotation matrix:")
print(Q_180)

print("\nDoes rotating twice equal 180° rotation?")
print(np.allclose(Q_twice, Q_180))

# Apply rotation twice to points
ROTATED_TWICE = TEST_POINTS @ Q_twice.T

print("\nOriginal -> Rotated twice")
for original, rotated in zip(TEST_POINTS, ROTATED_TWICE):
    print(
        f"({original[0]:7.1f}, {original[1]:7.1f})"
        f" -> "
        f"({rotated[0]:7.1f}, {rotated[1]:7.1f})"
    )
```

```python

# ============================================================
# SCREEN DIMENSION TRANSFORMATION
# ============================================================

portrait_width = PORTRAIT_WIDTH
portrait_height = PORTRAIT_HEIGHT

landscape_width = portrait_height
landscape_height = portrait_width

print("Screen dimensions")
print("-" * 40)

print(f"Portrait  : {portrait_width} × {portrait_height}")
print(f"Landscape : {landscape_width} × {landscape_height}")
```

```python

# ============================================================
# SCREEN CORNERS IN CENTERED COORDINATES
# ============================================================

w = PORTRAIT_WIDTH
h = PORTRAIT_HEIGHT

# Centered portrait screen corners
portrait_corners = np.array([
    [-w/2, -h/2],
    [ w/2, -h/2],
    [ w/2,  h/2],
    [-w/2,  h/2],
    [-w/2, -h/2]
])

# Rotate the corners
landscape_corners = portrait_corners @ Q.T

print("Portrait corners:")
print(portrait_corners)

print("\nRotated landscape corners:")
print(landscape_corners)
```

```python
# ============================================================
# SCREEN ROTATION VISUALIZATION
# ============================================================

fig, ax = plt.subplots(figsize=(11, 7))

# Portrait screen
ax.plot(
    portrait_corners[:, 0],
    portrait_corners[:, 1],
    marker='o',
    linewidth=2,
    label='Portrait'
)

# Landscape screen
ax.plot(
    landscape_corners[:, 0],
    landscape_corners[:, 1],
    marker='o',
    linewidth=2,
    label='After 90° rotation'
)

# Plot test points
ax.scatter(
    TEST_POINTS[:, 0],
    TEST_POINTS[:, 1],
    s=60,
    label='Original points'
)

ax.scatter(
    ROTATED_POINTS[:, 0],
    ROTATED_POINTS[:, 1],
    s=60,
    marker='x',
    label='Rotated points'
)

# Draw axes
ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

ax.set_xlabel("X coordinate")
ax.set_ylabel("Y coordinate")
ax.set_title("Screen Rotation as an Orthogonal Transformation")

ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.legend()

plt.show()
```

```python
# ============================================================
# SINGLE POINT ROTATION VISUALIZATION
# ============================================================

point = np.array([400, 300])
rotated_point = Q @ point

fig, ax = plt.subplots(figsize=(8, 8))

# Original vector
ax.arrow(
    0, 0,
    point[0], point[1],
    length_includes_head=True,
    head_width=35
)

# Rotated vector
ax.arrow(
    0, 0,
    rotated_point[0], rotated_point[1],
    length_includes_head=True,
    head_width=35
)

ax.scatter(point[0], point[1], s=80, label='Original point')
ax.scatter(
    rotated_point[0],
    rotated_point[1],
    s=80,
    marker='x',
    label='Rotated point'
)

ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("90° Rotation of a Screen Coordinate")

ax.grid(True, alpha=0.3)
ax.set_aspect('equal')
ax.legend()

plt.show()

print(f"Original point : {point}")
print(f"Rotated point  : {rotated_point}")
print(f"Original length: {np.linalg.norm(point):.4f}")
print(f"Rotated length : {np.linalg.norm(rotated_point):.4f}")
```

```python

# ============================================================
# BENCHMARK TEST DATASET
# ============================================================

benchmark_points = np.array([
    [100, 100],
    [200, 300],
    [-400, 250],
    [500, -200],
    [-300, -350],
    [0, 500]
], dtype=float)

benchmark_rotated = benchmark_points @ Q.T

benchmark_original_norms = np.linalg.norm(
    benchmark_points, axis=1
)

benchmark_rotated_norms = np.linalg.norm(
    benchmark_rotated, axis=1
)

benchmark_errors = np.abs(
    benchmark_original_norms - benchmark_rotated_norms
)

print("BENCHMARK RESULTS")
print("=" * 70)

print(
    f"{'Point':<8}"
    f"{'Original (x,y)':<22}"
    f"{'Rotated (x,y)':<22}"
    f"{'Length Error':>15}"
)

for i in range(len(benchmark_points)):
    print(
        f"{i+1:<8}"
        f"{str(benchmark_points[i]):<22}"
        f"{str(benchmark_rotated[i]):<22}"
        f"{benchmark_errors[i]:>15.10f}"
    )

print("\nMaximum benchmark error:",
      np.max(benchmark_errors))

print(
    "BENCHMARK PASSED:",
    np.allclose(
        benchmark_original_norms,
        benchmark_rotated_norms
    )
)
```

```python
# ============================================================
# FINAL PROJECT VERIFICATION
# ============================================================

orthogonal_test = np.allclose(Q.T @ Q, np.eye(2))

determinant_test = np.isclose(np.linalg.det(Q), 1)

distance_test = np.allclose(
    np.linalg.norm(TEST_POINTS, axis=1),
    np.linalg.norm(ROTATED_POINTS, axis=1)
)

quadratic_test = all(
    np.isclose(
        quadratic_form(TEST_POINTS[i]),
        quadratic_form(ROTATED_POINTS[i])
    )
    for i in range(len(TEST_POINTS))
)

double_rotation_test = np.allclose(
    Q @ Q,
    Q_180
)

print("=" * 65)
print("                 FINAL PROJECT RESULTS")
print("=" * 65)

print(f"1. Orthogonal matrix Q^T Q = I     : {'PASS' if orthogonal_test else 'FAIL'}")
print(f"2. Determinant det(Q) = +1         : {'PASS' if determinant_test else 'FAIL'}")
print(f"3. Distances preserved              : {'PASS' if distance_test else 'FAIL'}")
print(f"4. Quadratic form invariant         : {'PASS' if quadratic_test else 'FAIL'}")
print(f"5. Two rotations = 180° rotation   : {'PASS' if double_rotation_test else 'FAIL'}")

print("=" * 65)

all_tests = (
    orthogonal_test
    and determinant_test
    and distance_test
    and quadratic_test
    and double_rotation_test
)

if all_tests:
    print("FINAL STATUS: ALL MATHEMATICAL TESTS PASSED ✓")
else:
    print("FINAL STATUS: SOME TESTS FAILED")
```
